```python
copy=df.copy()
copy['Quantity']=df['Quantity'].fillna(df['Quantity'].mean())
copy['Discount']=df['Discount'].fillna(df['Discount'].median())
copy['PaymentMethod']=df['PaymentMethod'].fillna(df['PaymentMethod'].mode()[0])
copy['OrderDate']=df['OrderDate'].fillna(df['OrderDate'].ffill())
```

```python
missing_cleaned=copy.isnull().sum()
plt.bar(missing_cleaned.index,missing_cleaned.values)
plt.xlabel('columns')
plt.ylabel('missing values')
plt.title('missing values')
plt.xticks(rotation=45)
plt.show()
```
